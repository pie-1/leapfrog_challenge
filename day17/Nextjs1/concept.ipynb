# intercepting Routes
- it allows us to load a route from another part of an application within the current layout
- it can be useful when u want to display content of a route without user switching to a different context
- ex : when clicking on photo in a feed, you can display photo in a modal, overlaying the feed.in this case nextjs intercepts the /photo/123 route, masks the url and overlays it over /feed
  - however when navigating to photo by clicking a shareable URL or by refreshing page entire photo page should render instead of modal. No route interception should occur

## how to implement authentication in Next.js

- react and nextjs feature to use to implement auth
1. Authentication : verifies if the user is who they say they are, it requires the user to prove their identity with something they have such as a username or password
